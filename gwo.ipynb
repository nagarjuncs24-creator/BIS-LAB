```python

import numpy as np

# Generator data
PD = 300
Pmin = np.array([50, 50, 50])
Pmax = np.array([200, 150, 100])

a = np.array([100, 120, 150])
b = np.array([5.3, 5.5, 5.8])
c = np.array([0.004, 0.006, 0.009])

# Fitness function
def fitness(x):
    p = np.array([x[0], x[1], PD-x[0]-x[1]])

    if np.any(p < Pmin) or np.any(p > Pmax):
        return 1e10

    return np.sum(a + b*p + c*p**2)

# GWO initialization
N, T = 20, 100
wolves = np.random.uniform([50, 50], [200, 150], (N, 2))

for t in range(T):

    scores = np.array([fitness(x) for x in wolves])
    order = np.argsort(scores)

    alpha = wolves[order[0]].copy()
    beta = wolves[order[1]].copy()
    delta = wolves[order[2]].copy()

    best = scores[order[0]]
    A_factor = 2 - 2*t/T

    for i in range(N):
        X = wolves[i]

        positions = []

        for leader in [alpha, beta, delta]:
            A = 2*A_factor*np.random.rand(2) - A_factor
            C = 2*np.random.rand(2)

            D = np.abs(C*leader - X)
            positions.append(leader - A*D)

        wolves[i] = np.mean(positions, axis=0)

        # Keep generator outputs within limits
        wolves[i] = np.clip(wolves[i], [50, 50], [200, 150])

print("Optimal Generator Outputs:")

P1, P2 = alpha
P3 = PD - P1 - P2

print("Generator 1:", round(P1, 2), "MW")
print("Generator 2:", round(P2, 2), "MW")
print("Generator 3:", round(P3, 2), "MW")
print("Minimum Fuel Cost:", round(best, 2))
```
