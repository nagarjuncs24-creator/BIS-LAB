```python
import numpy as np

# Objective function
def fitness(x):
    return x**2 + 5*x + 6

# Parameters
N = 10
iterations = 20
pa = 0.25
step_size = 0.5

# Initialize nests
nests = np.random.uniform(-10, 10, N)
best = min(nests, key=fitness)

# Main algorithm
for iteration in range(iterations):

    for i in range(N):

        # Generate new solution using random movement
        new = nests[i] + np.random.normal(0, step_size)

        # Keep solution within limits
        new = np.clip(new, -10, 10)

        # Hill climbing selection
        if fitness(new) < fitness(nests[i]):
            nests[i] = new

        # Update global best
        if fitness(nests[i]) < fitness(best):
            best = nests[i]

    # Cuckoo Search: abandon worst nests
    count = int(pa * N)
    worst = np.argsort([fitness(x) for x in nests])[-count:]

    for i in worst:
        nests[i] = np.random.uniform(-10, 10)

    # Update best solution
    current = min(nests, key=fitness)
    if fitness(current) < fitness(best):
        best = current

    if (iteration + 1) % 10 == 0:
        print("Iteration:", iteration + 1,
              "Best Solution:", round(best, 5),
              "Fitness:", round(fitness(best), 5))

print("\nFinal Results")
print("Optimal Solution:", round(best, 5))
print("Minimum Fitness:", round(fitness(best), 5))
```
