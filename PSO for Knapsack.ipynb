```python
import random

# Knapsack data
profits = [60, 100, 120, 80, 90]
weights = [10, 20, 30, 15, 25]
capacity = 60

# PSO parameters
n = 50
iterations = 100

# Initialize particles
particles = [
    [random.randint(0, 1) for _ in profits]
    for _ in range(n)
]

velocities = [[0]*len(profits) for _ in range(n)]

pbest = [p[:] for p in particles]

def fitness(p):
    w = sum(p[i]*weights[i] for i in range(len(p)))
    v = sum(p[i]*profits[i] for i in range(len(p)))
    return v if w <= capacity else 0

gbest = max(pbest, key=fitness)[:]

# PSO algorithm
for _ in range(iterations):
    for i in range(n):
        for j in range(len(profits)):

            if fitness(particles[i]) > fitness(pbest[i]):
                pbest[i] = particles[i][:]

            if fitness(pbest[i]) > fitness(gbest):
                gbest = pbest[i][:]

            velocities[i][j] = (
                0.5*velocities[i][j]
                + 1.5*random.random()*(pbest[i][j]-particles[i][j])
                + 1.5*random.random()*(gbest[j]-particles[i][j])
            )

            prob = 1/(1+2.71828**(-velocities[i][j]))
            particles[i][j] = 1 if random.random() < prob else 0

        # Repair capacity
        while sum(particles[i][j]*weights[j] for j in range(len(profits))) > capacity:
            selected = [j for j in range(len(profits)) if particles[i][j] == 1]
            if not selected:
                break
            j = min(selected, key=lambda x: profits[x]/weights[x])
            particles[i][j] = 0

# Final result
best = gbest

print("Product/Item Selection:", best)
print("Maximum Profit:", fitness(best))
print("Total Weight Used:", sum(best[i]*weights[i] for i in range(len(best))))
```
